## <center> Main

#### **download conversion table**

```python
dataset = Dataset(
    name="hsapiens_gene_ensembl",
    host="http://www.ensembl.org",
)

annot = dataset.query(attributes=[
    "ensembl_gene_id",
    "external_gene_name",
    "gene_biotype",
    "chromosome_name",
])

annot = annot.rename(columns={
    "Gene stable ID": "gene_id",
    "Gene name": "gene_symbol",
    "Gene type": "gene_biotype",
    "Chromosome/scaffold name": "chromosome",
})

annot = annot.drop_duplicates("gene_id").set_index("gene_id")

annot.head()
```
