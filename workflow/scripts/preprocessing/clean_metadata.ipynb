## <center> Set-up

#### **libs**

```python
import pandas as pd
## helper functions
from workflow.scripts.funs.helper_funs import get_project_root
```

#### **paths**

```python
from pathlib import Path

if "snakemake" in globals():
    in_metadata = Path(snakemake.input["in_metadata"])
    out_metadata = Path(snakemake.output["out_metadata"])
else:
    ## get paths
    ROOT = get_project_root(globals())
    ## input
    in_metadata = ROOT / "data/metadata.csv"
    ## output
    out_metadata = (ROOT / "output/preprocessing/objects/clean_metadata/metadata.tsv")
    out_metadata.parent.mkdir(parents=True, exist_ok=True)
```

## <center> Main

```python
# load raw metadata table
meta = pd.read_csv(in_metadata, sep=',')

meta.head()
```

```python
# clean

## melt so that each sample is in one row
meta = meta.melt(id_vars=['disease_group', 'QC_group'], value_vars=['IFNG_neg','IFNG_pos'], var_name='condition_group', value_name='sample_id')
## remove space
meta['disease_group']=['HR_case' if x=='HR Case' else 'HR_control' for x in meta['disease_group']]
meta['QC_group']=['A' if x=='Group A' else 'B' for x in meta['QC_group']]

meta.head()
```

#### **save**

```python
meta.to_csv(out_metadata,sep='\t')
```
